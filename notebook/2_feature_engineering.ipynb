### Upcoming Feature Engineering

The following engineered features will be added to enhance customer behavioral and value analysis:

- **Ticket_Size** – Average transaction size of the customer.
- **Utilization** – Credit utilization ratio based on available credit vs usage.
- **Engagement** – Composite engagement metric derived from activity indicators.
- **Portfolio_Yield_Proxy** – Proxy indicator for revenue contribution from a customer.
- **CLV_Score** – Customer Lifetime Value score estimating long-term profitability.

**KPI_1: Ticket Size**
- Ticket Size = Total Transaction Amount / Transaction Count

```python
df["Ticket_Size"] = df["Total_Trans_Amt"] / df["Total_Trans_Ct"]
df["Ticket_Size"].describe()
```

```python
print("Interpreation")
print("Small ticket → everyday purchases")
print("Large ticket → occasional big purchases{Travel/Luxury}")

plt.figure(figsize=(5,3))
sns.histplot(df["Ticket_Size"], bins=50)
plt.title("Ticket Size Distribution")
plt.show()

print("Interpretation: Most customers have a ticket size between 100 and 500, indicating regular spending patterns. \nA long tail of higher ticket sizes suggests a smaller segment of customers making larger purchases.")
```

**KPI_2: Credit Utilization**
- Utilization = Revolving Balance / Credit Limit

```python
df["Utilization"] = df["Total_Revolving_Bal"] / df["Credit_Limit"]
df["Utilization"].describe()
```

```python
print("\nTypical Thresholds:")
print("<30%  : Healthy credit usage")
print("30–70% : Moderate utilization")
print(">70%  : Potential high-risk segment")

plt.figure(figsize=(5,3))
sns.histplot(df["Utilization"], bins=50)
plt.title("Credit Utilization Distribution")
plt.show()

print("\nInterpretation:")
print("Customers with utilization below 30% are generally considered financially healthy.")
print("Utilization between 30% and 70% indicates moderate credit usage.")
print("Utilization above 70% may signal higher financial risk or potential credit stress.")
```

**KPI_3: Engagement Score**
- Engagement = Transaction Count / Months on Book

```python
df["Engagement"] = df["Total_Trans_Ct"] / df["Months_on_book"]
df["Engagement"].describe()
```

```python
print("\nTypical Thresholds:")
print("Low Engagement     : Infrequent transactions or account activity")
print("Moderate Engagement: Regular usage of card for daily spending")
print("High Engagement    : Frequent transactions and active card usage")

plt.figure(figsize=(5,3))
sns.histplot(df["Engagement"], bins=50)
plt.title("Customer Engagement Distribution")
plt.show()

print("\nInterpretation:")
print("Customers with low engagement may represent dormant or inactive accounts.")
print("Moderately engaged customers typically use the card for routine spending.")
print("Highly engaged customers interact frequently and often represent valuable long-term customers.")
```

**Portfolio Yield Proxy**
- Portfolio Yield Proxy = Transaction Amount × Utilization

```python
df["Portfolio_Yield_Proxy"] = df["Total_Trans_Amt"] * df["Utilization"]
df["Portfolio_Yield_Proxy"].describe()
```

```python
print("\nTypical Thresholds:")
print("Low Yield       : Customers generating low interest or fee revenue")
print("Moderate Yield  : Customers contributing stable revenue to the portfolio")
print("High Yield      : Customers generating significant interest income or fees")

plt.figure(figsize=(5,3))
sns.histplot(df["Portfolio_Yield_Proxy"], bins=50)
plt.title("Portfolio Yield Proxy Distribution")
plt.xlabel("Portfolio_Yield_Proxy")
plt.ylabel("Count")
plt.show()

print("\nInterpretation:")
print("Customers with low portfolio yield contribute minimal revenue to the bank.")
print("Moderate yield customers represent stable and consistent revenue generators.")
print("High yield customers generate substantial interest or fee income and are often high-value segments.")
```

**Customer Lifetime Value(CLV) Approximation**
- CLV Score = Transaction Amount × (1 + Utilization) × Engagement

```python
df["CLV_Score"] = df["Total_Trans_Amt"] * (1 + df["Utilization"]) * df["Engagement"]
df["CLV_Score"].describe()
```

```python
print("\nTypical Thresholds:")
print("Low CLV        : Customers expected to generate limited long-term value")
print("Moderate CLV   : Customers with stable and consistent value contribution")
print("High CLV       : High-value customers expected to generate significant lifetime revenue")

plt.figure(figsize=(5,3))
sns.histplot(df["CLV_Score"], bins=50)
plt.title("CLV Score Distribution")
plt.xlabel("CLV_Score")
plt.ylabel("Count")
plt.show()

print("\nInterpretation:")
print("Customers with low CLV contribute limited long-term revenue to the business.")
print("Moderate CLV customers represent stable contributors to overall portfolio value.")
print("High CLV customers are the most valuable segment and should be prioritized for retention and premium offerings.")
```

**Transaction Velocity**

```python
df["Transaction_Velocity"] = df["Total_Trans_Ct"] / 12
df["Transaction_Velocity"].describe()
```

```python
print("\nTypical Thresholds:")
print("Low Velocity      : Infrequent transactions over time")
print("Moderate Velocity : Regular transaction activity")
print("High Velocity     : Very frequent transactions within short periods")

plt.figure(figsize=(5,3))
sns.histplot(df["Transaction_Velocity"], bins=50)
plt.title("Transaction Velocity Distribution")
plt.xlabel("Transaction_Velocity")
plt.ylabel("Count")
plt.show()

print("\nInterpretation:")
print("Customers with low transaction velocity use their card occasionally.")
print("Moderate velocity indicates consistent card usage for everyday spending.")
print("High velocity customers transact frequently and represent highly active users.")
```
